## test 1 - zamiana frazy na url


```python
from scraper.wikifetcher import WikiFetcher

fetcher = WikiFetcher("https://en.wikiquote.org/wiki/")
assert (fetcher._get_url("The Lord of the Rings") == "https://en.wikiquote.org/wiki/The_Lord_of_the_Rings")
print("TEST OK: URL poprawnie wygenerowany.")
```

## test 2 - liczba słów

```python
from scraper.wikiarticle import WikiArticle

file_path = "data/arwen_owtrta.html"

with open(file_path, "r", encoding="utf-8") as f:
    html_string = f.read()
    
article = WikiArticle(html_string, phrase="Arwen")
text=article.count_words()
if "word count: 2183" in text:
    print("TEST OK: Liczba słów poprawna.")
else:
    print("TEST FAILED: Liczba słów niepoprawna.")
```
